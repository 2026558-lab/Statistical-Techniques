```python
#Libraries 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats
```

```python
# Plot the original
df = pd.read_csv("online_shoppers_intention.csv")
df.head()
```

```python
# Plot the original
df.info()
```

```python
# Plot the original
df.isnull().sum()
```

```python
# Plot the original
print (df.describe().round(2))
```

```python
# Plot the original
print (df.describe(include="object"))
```

```python
# Plot the original
print (df['Revenue'].value_counts())
```

```python
# Plot the original
print ((df['Revenue'].value_counts(normalize=True)*100).round(2))
```

```python
# Plot the original purchase vs no purchase
sns.countplot(data=df, x="Revenue")
```

```python
# Plot the original uniform distribution Browsing Time
sns.histplot(df["ProductRelated_Duration"], bins=30, color='orange', edgecolor='black')
plt.title('Product-Related Browsing Time')
plt.show()
```

```python
# Plot the boxplot Browsing time and purchases 
sns.boxplot (data=df, X="Revenue", y="ProductRelated_Duration")
plt.title ('Browsing Time by Purchase Outcome')
plt.show()
```
